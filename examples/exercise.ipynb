```python
## SOLUTION
def sum_of_squares(n):
    total = 0
    for k in range(1, n + 1):
        total += k * k
    return total

sum_of_squares(10)
```

!! SOLUTION

**Explanation:** `total` starts at 0 and the loop adds $k^2$ for
$k = 1, \dots, n$. A shorter version is `sum(k * k for k in range(1, n + 1))`,
and the closed formula is $n(n+1)(2n+1)/6$.
